## 문제 1. 좌표의 사분면 판별
x, y 좌표를 입력받아 제1~제4사분면 중 어디에 있는지 `if-elif-else`로 판별합니다. 단, x=0 또는 y=0은 제외합니다.

```python
def quadrant(x, y):
    if x == 0 or y == 0:
        return 'x=0 또는 y=0은 문제 조건에서 제외됩니다.'
    elif x > 0 and y > 0:
        return '제1사분면'
    elif x < 0 and y > 0:
        return '제2사분면'
    elif x < 0 and y < 0:
        return '제3사분면'
    else:
        return '제4사분면'

x, y = -100, 2000
print(f'x={x}, y={y} -> {quadrant(x, y)}')
```

## 문제 2. 연도와 월을 입력받아 그 달의 마지막 날 출력
윤년 조건을 반영해 해당 월의 일수를 출력합니다.

```python
def is_leap_year(year):
    return year % 400 == 0 or (year % 4 == 0 and year % 100 != 0)

def last_day(year, month):
    if month in [1,3,5,7,8,10,12]:
        return 31
    elif month in [4,6,9,11]:
        return 30
    elif month == 2:
        return 29 if is_leap_year(year) else 28
    else:
        return None

for year, month in [(2008, 10), (2024, 2), (2100, 2), (2000, 2)]:
    print(year, month, '->', last_day(year, month))
```

## 문제 4. 랜덤 모듈과 함수를 이용한 가위바위보
사용자가 가위/바위/보를 선택하고, 컴퓨터는 랜덤으로 선택해 승패를 판정합니다.

```python
import random
random.seed(30801)

def rps(user):
    choices = ['가위', '바위', '보']
    computer = random.choice(choices)
    if user not in choices:
        result = '잘못된 입력'
    elif user == computer:
        result = '비김'
    elif (user, computer) in [('가위','보'), ('바위','가위'), ('보','바위')]:
        result = '사용자 승'
    else:
        result = '컴퓨터 승'
    return computer, result

user = '바위'
computer, result = rps(user)
print('사용자:', user)
print('컴퓨터:', computer)
print('결과:', result)
```

## 문제 5. 동물 이름 리스트 만들기
두 글자 이상의 동물 이름만 입력받고, 이미 있는 이름은 중복 추가하지 않습니다. 마지막에는 오름차순으로 정렬합니다.

```python
def add_animal(animal_list, name):
    name = name.strip()
    if name == '':
        return '입력 종료'
    if len(name) < 2:
        return '두 글자 이상의 동물 이름을 입력하세요.'
    if name in animal_list:
        return f'{name}은(는) 이미 들어있습니다.'
    animal_list.append(name)
    return f'{name} 추가'

animals = []
inputs = ['호랑이', '곰', '사자', '호랑이', '닭', '']
for name in inputs:
    msg = add_animal(animals, name)
    print(f'입력: {name!r} -> {msg}')
    if msg == '입력 종료':
        break
animals.sort()
print('최종:', animals)
```

## 문제 6. 10~99 사이 두 자리 숫자 맞히기
정답은 랜덤으로 정하고, 두 자리 숫자가 아닌 입력은 제외하며 맞힐 때까지 시도 횟수를 셉니다.

```python
random.seed(94)  # 실행 결과를 재현하기 위한 고정 시드
answer = random.randint(10, 99)
guesses = ['ab', '9', '90', '94', str(answer)]
count = 0
print('정답(검증용):', answer)
for g in guesses:
    print('입력:', g)
    if not g.isdigit() or not (10 <= int(g) <= 99):
        print('두 자리 숫자가 아닙니다.')
        continue
    count += 1
    value = int(g)
    if value == answer:
        print('맞혔습니다!')
        print(f'{count}번만에 맞혔어요.')
        break
    elif value < answer:
        print('더 큰 수를 입력하세요.')
    else:
        print('더 작은 수를 입력하세요.')
```

## 문제 8. 랜덤 이름·나이·e-mail·전화번호·성별 생성
각 값을 함수에서 생성하고 `return`으로 반환하여 결과를 출력합니다.

```python
import string
random.seed(30801)

last_names = ['김','이','박','최','정','강','조','윤','장','임']
first_names = ['도윤','서준','지호','현우','서연','지민','하린','수빈','예진','민재']

def random_name():
    return random.choice(last_names) + random.choice(first_names)

def random_age():
    return random.randint(20, 60)

def random_mail(length=8):
    chars = string.ascii_lowercase + string.digits
    user = ''.join(random.choice(chars) for _ in range(length))
    return user + '@dajeong.com'

def random_phone():
    return f"010-{random.randint(0,9999):04d}-{random.randint(0,9999):04d}"

def random_gender():
    return random.choice(['male', 'female'])

def make_person():
    return {
        'name': random_name(),
        'age': random_age(),
        'e-mail': random_mail(),
        'phone': random_phone(),
        'gender': random_gender()
    }

people = [make_person() for _ in range(10)]
for p in people:
    print(p)
```

## 문제 9. 생성 결과를 CSV 파일로 저장
사진의 CSV 예시와 연결되도록 문제 8에서 만든 10명의 결과를 `30801.csv`로 저장합니다.

```python
import csv

filename = '30801.csv'
with open(filename, 'w', newline='', encoding='utf-8-sig') as f:
    writer = csv.DictWriter(f, fieldnames=['name','age','e-mail','phone','gender'])
    writer.writeheader()
    writer.writerows(people)

print(filename, '저장 완료')
with open(filename, 'r', encoding='utf-8-sig') as f:
    print(f.read())
```

## 문제 10. 버블 정렬과 개선된 버블 정렬
인접한 두 값을 비교하여 오름차순으로 정렬합니다. 개선 버전은 한 회전에서 교환이 없으면 정렬이 끝났다고 판단해 즉시 종료합니다.

```python
def bubble_sort(arr):
    a = arr.copy()
    steps = []
    n = len(a)
    for i in range(n-1):
        for j in range(n-1-i):
            if a[j] > a[j+1]:
                a[j], a[j+1] = a[j+1], a[j]
        steps.append(a.copy())
    return a, steps

def improved_bubble_sort(arr):
    a = arr.copy()
    steps = []
    n = len(a)
    for i in range(n-1):
        swapped = False
        for j in range(n-1-i):
            if a[j] > a[j+1]:
                a[j], a[j+1] = a[j+1], a[j]
                swapped = True
        steps.append(a.copy())
        if not swapped:
            break
    return a, steps

array = [12,5,8,21,1,15,7,3,10,18]
print('원본:', array)
result, steps = bubble_sort(array)
for i, s in enumerate(steps, 1):
    print(f'단계{i}:', s)
print('버블정렬 결과:', result)

result2, steps2 = improved_bubble_sort(array)
print('개선 버블정렬 단계 수:', len(steps2))
print('개선 버블정렬 결과:', result2)
```

## 문제 11. 선택 정렬
기준 위치를 정하고, 뒤쪽 범위에서 가장 작은 값을 찾아 기준 위치와 교환하는 과정을 반복합니다.

```python
def selection_sort(arr):
    a = arr.copy()
    steps = []
    n = len(a)
    for i in range(n-1):
        min_idx = i
        for j in range(i+1, n):
            if a[j] < a[min_idx]:
                min_idx = j
        a[i], a[min_idx] = a[min_idx], a[i]
        steps.append(a.copy())
    return a, steps

array = [12,5,8,21,1,15,7,3,10,18]
print('원본:', array)
result, steps = selection_sort(array)
for i, s in enumerate(steps, 1):
    print(f'단계{i}:', s)
print('선택정렬 결과:', result)
```
